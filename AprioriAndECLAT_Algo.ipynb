```python
# convert this df into list of list
transactions=[]
for i in range(data.shape[0]):
    transactions.append([str(data.values[i,j]) for j in range(data.shape[1]) if str(data.values[i,j])!='nan'])
```

```python
transactions
```

```python
!pip install mlxtend
```

```python
# Transaction encoding
from mlxtend.preprocessing import TransactionEncoder
te=TransactionEncoder()
te_array=te.fit_transform(transactions)

df=pd.DataFrame(te_array,columns=te.columns_)
df
```

# Apriori algo

```python
from mlxtend.frequent_patterns import apriori,association_rules
freq_itemset=apriori(df,min_support=0.01,use_colnames=True)
freq_itemset.head()
```

```python
freq_itemset.shape
```

```python
# generate association rules
num_itemsets=len(freq_itemset)
rules=association_rules(freq_itemset,num_itemsets=num_itemsets,metric='confidence',min_threshold=0.2)
rules.head()
```

```python
rules.tail()
```

Generate strong rule

```python
strong_rules=rules[(rules['confidence']>0.4) | (rules['lift']>1.2)]
strong_rules
```

```python
strong_rules.shape
```

```python
# Example transaction data
transactions = {
    'T1': ['A', 'B', 'D'],
    'T2': ['B', 'C'],
    'T3': ['A', 'B', 'C', 'E'],
    'T4': ['B', 'E'],
    'T5': ['A', 'B', 'C', 'E']
}
```

```python
# Convert dictionary to DataFrame
transactions_df = pd.DataFrame.from_dict(transactions, orient='index').stack().reset_index(level=1, drop=True).to_frame(name='items')
transactions_df = transactions_df.reset_index().groupby('index')['items'].apply(list).reset_index(name='items')
transactions_df = pd.DataFrame(transactions_df['items'].tolist())
```

```python
transactions_df
```

```python
!pip install pyECLAT
from pyECLAT import ECLAT
```

```python
# Fit ECLAT
eclat = ECLAT(data=transactions_df)
frequent_itemsets = eclat.fit(min_support=0.4, min_combination=1, max_combination=3)
print("Frequent itemsets with support:\n", frequent_itemsets)
```

```python
frequent_itemsets[1]
```

```python
import pandas as pd
from mlxtend.frequent_patterns import association_rules

# The frequent itemsets provided by the user are in the second element of the tuple
frequent_itemsets_dict = frequent_itemsets[1]

# Convert the dictionary to a pandas DataFrame
freq_itemset_df = pd.DataFrame(list(frequent_itemsets_dict.items()), columns=['itemsets', 'support'])

# Convert the 'itemsets' column to frozensets to match the expected format for association_rules
freq_itemset_df['itemsets'] = freq_itemset_df['itemsets'].apply(lambda x: frozenset(x.split(' & ')))


# Generate association rules using mlxtend
rules_from_eclat = association_rules(freq_itemset_df, metric='confidence', min_threshold=0.7)

# Display the generated rules
display(rules_from_eclat)
```
